# [5] Evaluation & Future Steps

### 5.1 Identified Model Limitations & Potential Flaws
1. **Unexplained Variance (~45%–50% Residual Variance):**
   - Important behavioral and clinical confounders — such as **HbA1c (glycated hemoglobin), medication dosage/adherence, physical activity, dietary intake, and genetic predisposition** — are not captured in the 10 baseline predictors.
2. **Mild Heteroscedasticity in Mid-to-High Predictions:**
   - Residual error dispersion widens slightly for predicted progression scores between `140` and `230`, partially violating the constant-variance assumption of Ordinary Least Squares.
3. **Single-Snapshot Cross-Sectional Predictors:**
   - All predictors were measured only once at baseline, preventing the model from capturing longitudinal trajectories (e.g., whether a patient's `BMI` was rising or falling prior to baseline).

### 5.2 Recommended Future Improvements & Alternative Techniques
1. **Target Transformation (`Box-Cox` / Log-Link Generalized Linear Model):**
   - Apply a logarithmic or Box-Cox transformation (or Gamma GLM) to `Disease_Progression` to stabilize residual variance and guarantee strictly positive predictions.
2. **Collect Longitudinal & Clinical Biomarkers (`HbA1c` & Continuous Glucose Monitoring):**
   - Integrate longitudinal `HbA1c` records, insulin treatment history, and lifestyle data to reduce residual error (`RMSE`).
3. **Quantile Regression & Non-Linear Ensemble Benchmarks:**
   - Train **Quantile Regression** (10th, 50th, 90th percentiles) to provide clinicians with calibrated **prediction intervals**, and benchmark against **Gradient Boosted Trees coupled with SHAP values** to capture complex non-linear interactions.
