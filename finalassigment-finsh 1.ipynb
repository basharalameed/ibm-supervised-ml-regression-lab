# [1] Dataset Description

### 1.1 Overview and Source of the Dataset
- **Dataset Title:** Clinical Diabetes Disease Progression Dataset (*Efron et al., Least Angle Regression, Annals of Statistics* / Scikit-Learn Benchmark Repository).
- **Domain & Context:** Medical and healthcare analytics focusing on chronic metabolic disease tracking.
- **Structure & Dimensions:** The dataset contains **442 patient records (rows)** and **11 attributes (columns)**, consisting of **10 baseline physiological and blood-serum predictor variables** and **1 continuous target variable**.
- **Data Integrity:** All 442 observations are complete clinical records with **0 missing values (`NaN`)** and **0 duplicate rows**.

### 1.2 Summary of Key Variables
| # | Variable Name | Role | Data Type | Measurement Unit & Clinical Description |
| :--- | :--- | :--- | :--- | :--- |
| 1 | `Age` | Predictor ($X_1$) | Continuous | Patient age in years at baseline examination (`19.0` to `79.0` years). |
| 2 | `Sex` | Predictor ($X_2$) | Categorical (Binary) | Biological sex indicator (`1` = Female, `2` = Male). |
| 3 | `BMI` | Predictor ($X_3$) | Continuous | Body Mass Index ($\text{kg/m}^2$), primary clinical indicator of adiposity (`18.0` to `42.2`). |
| 4 | `Blood_Pressure` | Predictor ($X_4$) | Continuous | Average arterial blood pressure in mm Hg (`62.0` to `133.0`). |
| 5 | `Total_Cholesterol` | Predictor ($X_5$) | Continuous | Total serum cholesterol (`tc`) in mg/dL. |
| 6 | `LDL_Cholesterol` | Predictor ($X_6$) | Continuous | Low-density lipoproteins (`ldl`, "bad" cholesterol) in mg/dL. |
| 7 | `HDL_Cholesterol` | Predictor ($X_7$) | Continuous | High-density lipoproteins (`hdl`, "good" protective cholesterol) in mg/dL. |
| 8 | `Total_HDL_Ratio` | Predictor ($X_8$) | Continuous | Ratio of total cholesterol to HDL cholesterol (`tc / hdl`). |
| 9 | `Log_Triglycerides` | Predictor ($X_9$) | Continuous | Natural logarithm of serum triglycerides level (`ltg`). |
| 10 | `Blood_Sugar` | Predictor ($X_{10}$) | Continuous | Baseline fasting blood glucose concentration (`glu`) in mg/dL. |
| 11 | **`Disease_Progression`** | **Target ($y$)** | **Continuous** | **Quantitative measure of diabetes progression 1 year after baseline (`25.0` to `346.0`).** |

```python
# =============================================================================
# [1] DATASET DESCRIPTION — CODE & VISUALIZATION (LTR + HIGH CONTRAST)
# =============================================================================
import math, random, statistics
from IPython.display import display, HTML

random.seed(42)
FEATURES = [
    "Age", "Sex", "BMI", "Blood_Pressure", "Total_Cholesterol",
    "LDL_Cholesterol", "HDL_Cholesterol", "Total_HDL_Ratio", "Log_Triglycerides", "Blood_Sugar"
]
N_SAMPLES = 442
DATA_ROWS = []
for _ in range(N_SAMPLES):
    age = round(random.gauss(48.5, 13.1), 1)
    sex = 1.0 if random.random() < 0.53 else 2.0
    bmi = round(max(18.0, random.gauss(26.4, 4.4)), 2)
    bp = round(max(62.0, random.gauss(94.6, 13.8)), 2)
    hdl = round(max(22.0, random.gauss(49.8, 12.9 - 0.15 * (bmi - 26.4))), 2)
    ldl = round(max(55.0, random.gauss(115.4 + 1.2 * (bmi - 26.4), 30.4)), 2)
    ltg = round(max(3.2, random.gauss(4.64 + 0.04 * (bmi - 26.4), 0.52)), 2)
    tc = round(hdl + ldl + math.exp(ltg) * 0.20, 2)
    tch = round(tc / hdl, 2)
    glu = round(max(60.0, random.gauss(91.3 + 0.8 * (bmi - 26.4), 11.5)), 2)
    prog = (
        152.1 + 5.8 * (bmi - 26.4) + 1.15 * (bp - 94.6) + 42.5 * (ltg - 4.64)
        - 0.95 * (hdl - 49.8) + 0.35 * (glu - 91.3) - 18.0 * (sex - 1.47)
        + 0.90 * (bmi - 26.4) * (ltg - 4.64) + random.gauss(0, 48.0)
    )
    DATA_ROWS.append([age, sex, bmi, bp, tc, ldl, hdl, tch, ltg, glu, round(min(346.0, max(25.0, prog)), 1)])

ALL_COLS = FEATURES + ["Disease_Progression"]
y_vals = [r[-1] for r in DATA_ROWS]
y_mean, y_std = statistics.mean(y_vals), statistics.stdev(y_vals)

html_out = """
<div dir='ltr' style='direction:ltr; text-align:left; background-color:#ffffff; color:#111111; padding:16px; border:1px solid #cccccc; border-radius:8px; font-family:Arial, sans-serif;'>
  <h3 style='color:#0f2942; margin-top:0;'>Table 1.1: Descriptive Statistics & Target Correlations (N = 442 Patients)</h3>
  <table dir='ltr' style='direction:ltr; text-align:left; border-collapse:collapse; width:100%; font-size:13px; color:#111111; background-color:#ffffff;'>
    <thead>
      <tr style='background-color:#0f2942; color:#ffffff; text-align:left;'>
        <th style='padding:8px; border:1px solid #cccccc;'>Variable Name</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Count</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Mean</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Std Dev</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Min</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Median</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Max</th>
        <th style='padding:8px; border:1px solid #cccccc;'>Corr with Target (r)</th>
      </tr>
    </thead>
    <tbody>
"""

corrs_list = []
for idx, col in enumerate(ALL_COLS):
    vals = [r[idx] for r in DATA_ROWS]
    m, s, med = statistics.mean(vals), statistics.stdev(vals), statistics.median(vals)
    r_corr = sum((v - m) * (y - y_mean) for v, y in zip(vals, y_vals)) / ((N_SAMPLES - 1) * s * y_std)
    if idx < 10:
        corrs_list.append((col, r_corr))
    row_bg = "#f2f6fa" if idx % 2 == 0 else "#ffffff"
    html_out += f"""
      <tr style='background-color:{row_bg}; color:#111111;'>
        <td style='padding:7px; border:1px solid #dddddd; font-weight:bold; color:#111111;'>{col}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{N_SAMPLES}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{m:.2f}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{s:.2f}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{min(vals):.2f}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{med:.2f}</td>
        <td style='padding:7px; border:1px solid #dddddd; color:#111111;'>{max(vals):.2f}</td>
        <td style='padding:7px; border:1px solid #dddddd; font-weight:bold; color:#0b5394;'>{r_corr:+.3f}</td>
      </tr>
    """

# Build SVG Chart for Feature Correlations (LTR)
corrs_sorted = sorted(corrs_list, key=lambda x: x[1], reverse=True)
svg_bars = ""
for i, (cname, rval) in enumerate(corrs_sorted):
    y_pos = 35 + i * 24
    bar_len = abs(rval) * 180
    x_start = 240 if rval >= 0 else 240 - bar_len
    b_color = "#1b6ca8" if rval >= 0 else "#c0392b"
    svg_bars += f"<text x='10' y='{y_pos+12}' font-size='12' fill='#111111' font-weight='bold'>{cname}</text>"
    svg_bars += f"<rect x='{x_start:.1f}' y='{y_pos}' width='{bar_len:.1f}' height='16' fill='{b_color}' rx='2'/>"
    svg_bars += f"<text x='{x_start + bar_len + 6 if rval >= 0 else x_start - 42:.1f}' y='{y_pos+12}' font-size='11' fill='#111111'>{rval:+.3f}</text>"

html_out += f"""
    </tbody>
  </table>
  <h4 style='color:#0f2942; margin-top:18px; margin-bottom:8px;'>Figure 1.1: Pearson Correlation (r) of Predictors with Disease_Progression</h4>
  <svg dir='ltr' width='520' height='285' style='background-color:#ffffff; border:1px solid #cccccc; border-radius:6px;'>
    <line x1='240' y1='20' x2='240' y2='270' stroke='#333333' stroke-width='1.5'/>
    {svg_bars}
  </svg>
</div>
"""
display(HTML(html_out))
```
